## Key performance indicators

```python
kpis = {
    'Total Sales': round(df['Sales'].sum(), 2),
    'Total Profit': round(df['Profit'].sum(), 2),
    'Total Orders': df['OrderID'].nunique(),
    'Units Sold': int(df['Quantity'].sum()),
    'Profit Margin %': round(df['Profit'].sum() / df['Sales'].sum() * 100, 2)
}
pd.Series(kpis)
```

## Sales by category

```python
category_sales = df.groupby('Category', as_index=False)['Sales'].sum().sort_values('Sales', ascending=False)
category_sales
```

```python
category_sales.plot(kind='bar', x='Category', y='Sales', legend=False, title='Sales by Category', rot=35)
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```

## Monthly sales trend

```python
df['OrderDate'] = pd.to_datetime(df['OrderDate'])
monthly_sales = df.assign(Month=df['OrderDate'].dt.to_period('M').astype(str)).groupby('Month', as_index=False)['Sales'].sum()
monthly_sales
```

```python
monthly_sales.plot(kind='line', x='Month', y='Sales', marker='o', title='Monthly Sales Trend', legend=False, rot=45)
plt.tight_layout()
plt.show()
```

## Top 10 products

```python
top_products = df.groupby('ProductName', as_index=False)['Sales'].sum().nlargest(10, 'Sales')
top_products
```

```python
top_products.sort_values('Sales').plot(kind='barh', x='ProductName', y='Sales', legend=False, title='Top Products by Sales')
plt.tight_layout()
plt.show()
```
